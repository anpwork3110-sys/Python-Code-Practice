Functions of TUPLE Unpacking

```python
stock_price=[('Titan',2000),('BSE',3500),('MCX',1600)]
```

```python
for share,price in stock_price: # this normal tuple unpacking using for loop
  print(share)
```

```python
def stock_check(stock_price): # this is method of tuple unpacking using function in output also we can get tuple using function.
  ltp=0
  company_name=''

  for share,price in stock_price:
    if price > ltp:
      ltp=price
      company_name=share
    else:
      pass
  return (company_name,ltp)
```

```python
type(stock_check(stock_price))
```

```python
stock_check(stock_price)
```

```python
share,price=stock_check(stock_price) # ouptut generated from tuple also can be unpacked in this way by assigning value but
                                     # first check tuple structure
```
